**Montevideo**

```python
import requests
from bs4 import BeautifulSoup
import json
```

```python
url = "https://www.infocasas.com.uy/venta/casas/montevideo"

headers = {
    "User-Agent": "Mozilla/5.0"
}

response = requests.get(url, headers=headers)

soup = BeautifulSoup(response.content, "html.parser")
```

```python
ubicaciones = soup.find_all("strong", class_="lc-location")[:10]

enlaces_propiedades = []

for ubicacion in ubicaciones:
    enlace = ubicacion.find_parent("a")
    enlaces_propiedades.append(enlace)
```

```python
precios = []

for enlace in enlaces_propiedades:
    precio = enlace.find("p", class_="main-price").get_text(strip=True)
    precio = precio.replace("U$S", "")
    precio = precio.replace(".", "")
    precio = precio.strip()
    precio = int(precio)
    precios.append(precio)

precios

```

```python
tamaños = []

for enlace in enlaces_propiedades:
    for elemento in enlace.find_all("strong"):
        if "m²" in elemento.get_text():
            tamaño = elemento.get_text(strip=True)
            tamaño = tamaño.replace("m²", "").strip()
            tamaño = int(tamaño)
            tamaños.append(tamaño)
            break

tamaños
```

```python
habitaciones = []

for enlace in enlaces_propiedades:
    for elemento in enlace.find_all("strong"):
        if"Dorm" in elemento.get_text():
            habitacion = elemento.get_text(strip=True)
            habitacion = habitacion.split()[0]
            habitacion = int(habitacion)
            habitaciones.append(habitacion)
            break

habitaciones
```

```python
links = []

for enlace in enlaces_propiedades:
    link = enlace["href"]
    link = "https://www.infocasas.com.uy" + link
    links.append(link)

links
```

```python
propiedades_montevideo = []

for i in range(10):
    propiedad = {
        "precio": precios[i],
        "tamaño": tamaños[i],
        "habitaciones": habitaciones[i],
        "link": links[i]
    }

    propiedades_montevideo.append(propiedad)

propiedades_montevideo
```

**Salto**

```python
url_salto = "https://www.infocasas.com.uy/venta/casas/salto"

response_salto = requests.get(url_salto, headers=headers)

soup_salto = BeautifulSoup(response_salto.content, "html.parser")

enlaces_salto = soup_salto.find_all("a", class_="lc-data")[:10]

precios_salto = []
tamaños_salto = []
habitaciones_salto = []
links_salto = []

for enlace in enlaces_salto:
    precio = enlace.find("p", class_="main-price").get_text(strip=True)
    precio = precio.replace("U$S", "")
    precio = precio.replace(".", "")
    precio = precio.strip()
    precio = int(precio)
    precios_salto.append(precio)

    tamaño = None
    habitacion = None

    for elemento in enlace.find_all("strong"):
        texto = elemento.get_text(strip=True)

        if "m²" in texto:
            tamaño = texto.replace("m²", "").strip()
            tamaño = int(tamaño)

        if "Dorm" in texto:
            habitacion = texto.split()[0]
            habitacion = int(habitacion)

    tamaños_salto.append(tamaño)
    habitaciones_salto.append(habitacion)

    link = enlace["href"]
    link = "https://www.infocasas.com.uy" + link
    links_salto.append(link)

propiedades_salto = []

for i in range(10):
    propiedad = {
        "precio": precios_salto[i],
        "tamaño": tamaños_salto[i],
        "habitaciones": habitaciones_salto[i],
        "link": links_salto[i]
    }
    propiedades_salto.append(propiedad)

propiedades_salto
```
